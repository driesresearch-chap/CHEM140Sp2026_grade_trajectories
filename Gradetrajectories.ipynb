```python
import matplotlib.pyplot as plt

# Define the units to be plotted, as requested by the user
selected_units = ['Unit 3', 'Unit 4', 'Unit 7', 'Final']

plt.figure(figsize=(10, 6))
for _, row in df.iterrows():
    plt.plot(selected_units, row[selected_units], color=color_map[row['Final Grade']], alpha=0.6)

plt.title('Overlayed Cohort Mastery Trajectories (Selected Units)')
plt.xlabel('Course Units')
plt.ylabel('Cumulative Spells Mastered')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
grade_groups = {
    'A': 'A/A-', 'A-': 'A/A-',
    'B+': 'B+/B/B-', 'B': 'B+/B/B-', 'B-': 'B+/B/B-',
    'C+': 'C+/C/C-', 'C': 'C+/C/C-', 'C-': 'C+/C/C-',
    'D+': 'D+/D/D-', 'D': 'D+/D/D-', 'D-': 'D+/D/D-',
    'F': 'F'
}

df['Grade Group'] = df['Final Grade'].map(grade_groups)

# Define the units to average
units_to_average = ['Unit 3', 'Unit 4', 'Unit 5', 'Unit 6', 'Unit 7', 'Final']

# Calculate the average spells for each grade group across all units
average_spells_by_grade = df.groupby('Grade Group')[units_to_average].mean()
display(average_spells_by_grade)
```

```python
plt.figure(figsize=(10, 6))

# Map grade groups to the base colors for plotting
plot_color_map = {
    'A/A-': 'blue',
    'B+/B/B-': 'green',
    'C+/C/C-': 'gold',
    'D+/D/D-': 'orange',
    'F': 'red'
}

for grade_group, row in average_spells_by_grade.iterrows():
    plt.plot(units_to_average, row[units_to_average], color=plot_color_map[grade_group], marker='o', label=grade_group)

plt.title('Average Spells Mastered by Grade Group Across Units')
plt.xlabel('Course Units')
plt.ylabel('Average Cumulative Spells Mastered')
plt.grid(True)
plt.legend(title='Final Grade Group')
plt.tight_layout()
plt.show()
```
